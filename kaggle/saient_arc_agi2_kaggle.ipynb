```python
import base64,io,zipfile,sys,pathlib,os
# Encoded ZIP contains only four published Python source files; no network dependency.
BUNDLE_BASE64 = ""
root=pathlib.Path(os.environ.get('SAIENT_ARC_KAGGLE_WORKING','/kaggle/working'))/'saient'
root.mkdir(parents=True,exist_ok=True)
with zipfile.ZipFile(io.BytesIO(base64.b64decode(BUNDLE_BASE64))) as z:
    expected={'arc_gremlin/__init__.py','arc_gremlin/saient_solver.py','arc_gremlin/spatial_programs.py','arc_gremlin/extended_solver.py'}
    assert set(z.namelist())==expected
    for name in sorted(expected):
        target=root/name
        target.parent.mkdir(parents=True,exist_ok=True)
        target.write_bytes(z.read(name))
sys.path.insert(0,str(root))
print('Loaded public Saient ARC Gremlin solver (offline).')
```

```python
import os,glob,json,pathlib
input_root=pathlib.Path(os.environ.get('SAIENT_ARC_KAGGLE_INPUT_ROOT','/kaggle/input'))
candidates=sorted(input_root.rglob('arc-agi_test_challenges.json'))
if len(candidates)!=1:
    raise RuntimeError(f'Expected exactly one ARC-AGI-2 test challenges JSON, found {len(candidates)} under {input_root}')
tasks=json.loads(candidates[0].read_text())
assert isinstance(tasks,dict) and tasks
print('Input:',candidates[0].name,'Tasks:',len(tasks))
```

```python
from arc_gremlin.saient_solver import solve_with_report,apply
from collections import Counter
def rotate(g):
    return [list(r) for r in zip(*g[::-1])]
def unrotate(g):
    return [list(r) for r in zip(*g)][::-1]
def recolour(g,mapping):
    return [[mapping.get(v,v) for v in row] for row in g]
def three_views(task):
    counts=Counter(c for ex in task['train'] for kind in ('input','output') for row in ex[kind] for c in row if c)
    counts.update(c for ex in task['test'] for row in ex['input'] for c in row if c)
    palette=sorted(counts,key=lambda c:(-counts[c],c))
    mapping={0:0,**{c:i+1 for i,c in enumerate(palette)}}
    reverse={v:k for k,v in mapping.items()}
    variants=[
        ('original',lambda g:g,lambda g:g),
        ('rotate90',rotate,unrotate),
        ('palette',lambda g:recolour(g,mapping),lambda g:recolour(g,reverse))]
    rank=[]
    for idx,(name,forward,inverse) in enumerate(variants):
        transformed={'train':[{'input':forward(ex['input']),'output':forward(ex['output'])} for ex in task['train']],
                     'test':[{'input':forward(ex['input'])} for ex in task['test']]}
        try:
            report=solve_with_report(transformed)
            hyp=report.get('best_hypothesis')
            tr=[apply(hyp,ex['input']) for ex in transformed['train']]
            exact=sum(a==ex['output'] for a,ex in zip(tr,transformed['train']))
            pixel=0.
            for a,ex in zip(tr,transformed['train']):
                b=ex['output']
                if len(a)==len(b) and a and len(a[0])==len(b[0]):
                    pixel+=sum(a[y][x]==b[y][x] for y in range(len(b)) for x in range(len(b[0])))/(len(b)*len(b[0]))
            pred=[inverse(g) for g in report['predictions']]
            assert len(pred)==len(task['test'])
            rank.append((exact,pixel/max(len(tr),1),-idx,name,pred))
        except Exception as err:
            print('Solver view failed:',name,type(err).__name__)
    rank.sort(reverse=True)
    if not rank:
        return [[ex['input'] for ex in task['test']]]*2
    first=rank[0][4]
    second=next((r[4] for r in rank[1:] if r[4]!=first),first)
    return first,second
def valid_grid(g):
    return isinstance(g,list) and 1<=len(g)<=30 and isinstance(g[0],list) and 1<=len(g[0])<=30 and all(isinstance(row,list) and len(row)==len(g[0]) and all(type(v) is int and 0<=v<=9 for v in row) for row in g)
def sanitized(g,fallback):
    return g if valid_grid(g) else fallback
submission={}
for index,(task_id,raw) in enumerate(tasks.items(),1):
    # test outputs never enter the solver
    clean={'train':[{'input':e['input'],'output':e['output']} for e in raw['train']],
           'test':[{'input':e['input']} for e in raw['test']]}
    try:
        first,second=three_views(clean)
    except Exception as err:
        print('Task fallback:',task_id,type(err).__name__)
        first=second=[e['input'] for e in clean['test']]
    rows=[]
    for j,example in enumerate(clean['test']):
        fallback=example['input']
        rows.append({'attempt_1':sanitized(first[j],fallback),'attempt_2':sanitized(second[j],fallback)})
    submission[task_id]=rows
    if index%40==0: print('Completed',index,'/',len(tasks))
assert set(submission)==set(tasks)
assert all(len(submission[k])==len(tasks[k]['test']) for k in tasks)
assert all(set(a)=={'attempt_1','attempt_2'} and valid_grid(a['attempt_1']) and valid_grid(a['attempt_2']) for arr in submission.values() for a in arr)
output=pathlib.Path(os.environ.get('SAIENT_ARC_KAGGLE_OUTPUT','/kaggle/working/submission.json'))
output.parent.mkdir(parents=True,exist_ok=True)
output.write_text(json.dumps(submission,separators=(',',':')))
print('Wrote:',output,'tasks:',len(submission),'test outputs:',sum(map(len,submission.values())))
```
